## 2. Exploratory Visualizations

We generate five diagnostic figures:
1. **Class Balance:** Distribution of target (0 = no disease, 1 = presence of disease).
2. **Numeric Feature Histograms:** Stratified by diagnostic outcome.
3. **Categorical Feature Frequencies:** Stratified by diagnostic outcome.
4. **Pearson Correlation Heatmap:** Identifying feature interdependencies.
5. **Numeric Feature Boxplots:** Outliers and quartile distributions.

```python
figs = eda.run(config.FIGURES_DIR, df)
for p in figs:
    if p.suffix == '.png':
        print(f'Generated: {p.name}')
        display(Image(filename=str(p)))
```

## 3. Key Observations & Clinical Insights

- **Class Balance:** 164 patients (54.1%) without heart disease vs 139 patients (45.9%) with heart disease. The dataset is well-balanced.
- **Missing Values:** Missing values occur only in `ca` (4 rows, 1.3%) and `thal` (2 rows, 0.7%).
- **Key Predictors:**
  - `thalach` (maximum heart rate) is significantly lower in heart disease patients (median ~142 bpm vs ~158 bpm in healthy subjects).
  - `oldpeak` (ST depression) is markedly higher in heart disease patients (median ~1.4 mm vs ~0.0 mm in healthy subjects).
  - `cp=4` (asymptomatic chest pain) and `exang=1` (exercise-induced angina) show strong positive correlation with disease.
- **Correlation:** Pairwise correlations between predictors remain below 0.60, confirming suitability for classification without multicollinearity issues.
